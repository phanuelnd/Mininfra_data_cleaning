```python
import pandas as pd

gasabo_data_path = "Gasabo_buildings_with_upi.csv"
df = pd.read_csv(gasabo_data_path)

# Count occurrences of each UPI
upi_counts = df['upi'].value_counts()

# Filter UPIs that appear more than once
duplicated_upis = upi_counts[upi_counts > 1]

# Get all duplicated rows
duplicated_rows = df[df['upi'].isin(duplicated_upis.index)].copy()

# Add the count to each duplicated row
duplicated_rows['upi_occurrences'] = duplicated_rows['upi'].map(upi_counts)

# Save the duplicated rows with counts
duplicated_rows.to_csv("duplicated_upi_rows_sample.csv", index=False)

# Prepare a separate stats DataFrame
stats_df = duplicated_upis.reset_index()
stats_df.columns = ['upi', 'count']

# Calculate total number of unique UPIs that appear more than once
total_duplicated_upis = stats_df.shape[0]

# Insert total_duplicated_upis as the first row in the stats DataFrame
stats_with_total = pd.concat(
    [pd.DataFrame([{'upi': 'TOTAL_DUPLICATED_UPI', 'count': total_duplicated_upis}]), stats_df],
    ignore_index=True
)

# Save the stats with total_duplicated_upis at the top
stats_with_total.to_csv("duplicated_upi_sample_stats.csv", index=False)

print("Duplicated rows saved to: duplicated_upi_rows_sample.csv")
print("UPI duplication stats saved to: duplicated_upi_sample_stats.csv")
```
