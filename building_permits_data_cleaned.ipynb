```python
import pandas as pd
import numpy as np
import altair as alt
import matplotlib.pyplot as plt
import tabulate 
```

```python
old_buiilding_permit_data  =  "BPMIS_DATA.xlsx"
new_buiilding_permit_data = "applications.xlsx"

old_building_permit_data = pd.read_excel(old_buiilding_permit_data)
new_building_permit_data = pd.read_excel(new_buiilding_permit_data)

# print(old_building_permit_data.head())
# print(new_building_permit_data.head())

# print(old_building_permit_data.columns)
# print("--------------------------------")
# print(new_building_permit_data.columns)

# Print each column name and the number of missing values in that column, using tabulate for a clear view

old_missing = list(zip(old_building_permit_data.columns, old_building_permit_data.isnull().sum()))
print(tabulate.tabulate(old_missing, headers=["Column Name", "Missing Values"], tablefmt="grid"))
print("-------------New Building Permits Data-------------------")
new_missing = list(zip(new_building_permit_data.columns, new_building_permit_data.isnull().sum()))
print(tabulate.tabulate(new_missing, headers=["Column Name", "Missing Values"], tablefmt="grid"))
```
